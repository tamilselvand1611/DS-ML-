```python
import matplotlib.pyplot as plt
from scipy.stats import binom
import numpy as np
```

```python
'''Placement Test
A student has a 70% chance of answering a coding question correctly.
There are 15 questions in the test.
Find the probability that the student answers exactly 12 questions correctly.'''
n=15
k=12
p=0.7
p1=binom.pmf(k,n,p)
print(p1)
p2=binom.cdf(k,n,p)
print(p2)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="lightblue",edgecolor="black",label="Binomial PMF")
plt.bar(k,p1,color="red",edgecolor="black",label=f"k = {k}")
plt.xlabel("Number of Qns answered")
plt.ylabel("Probability")
plt.title(f"Binomial distribution")
plt.legend()
plt.grid(axis="y",alpha=0.6)
plt.show()
```

```python
'''Cricket Example
A batsman has a 60% chance of hitting a boundary in an over.
He plays 10 overs.
Find the probability that he hits boundaries in exactly 7 overs.'''
n=10
k=7
p=0.6
pos=binom.pmf(k,n,p)
print(pos)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="yellow",edgecolor="black",label="Binomial PMF")
plt.bar(k,pos,color="lightgreen",edgecolor="black",label=f"k = {k}")
plt.xlabel("Number of Boundaries")
plt.ylabel("Probability")
plt.title(f"Binomial Distribution")
plt.legend()
plt.grid(axis="y",alpha=0.6)
plt.show()
```

```python
'''YouTube Ad Clicks
A YouTube ad has a 20% chance of being clicked by a viewer.
If 30 viewers watch the ad, find the probability that exactly 5 viewers click it.'''
n=30
k=5
p=0.2
pos=binom.pmf(k,n,p)
print(pos)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="blue",edgecolor="black",label="Binomial PMF")
plt.bar(k,pos,color="gold",edgecolor="black",label=f"k = {k}")
plt.title(f"Binomial Distribution")
plt.xlabel("No of viewers")
plt.ylabel("Probability")
plt.legend()
plt.grid(axis="y",alpha=0.4)
plt.show()
```
