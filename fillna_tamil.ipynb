```python
import pandas as pd
data = { "Age": [22, None, 29, None, 31],"City": ["Delhi", None, "Mumbai", None, "Chennai"],"Salary": [50000, 60000, None, None, 75000]}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.fillna(0))
```

```python
print(df["Age"].fillna(df["Age"].mean()))
```

```python
print(df["Age"].fillna(df["Age"].median()))
```

```python
print(df["City"].fillna(df["City"].mode()[0]))
```

```python
df.ffill()
```

```python
df.bfill()
```

```python
print(df.fillna(0, inplace=True))
```
