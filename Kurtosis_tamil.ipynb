```python
import numpy as np
import pandas as pd
from scipy.stats import kurtosis as k
data=pd.read_csv("1000_companies.csv")
data
```

```python
profit=data["Profit"]
profit
```

```python
kurto=k(profit,fisher=True)
max_val=max(profit)
min_val=min(profit)
mean_val=np.mean(profit)
print("Kurosis :",kurto)
if kurto>0:
    print("More distributions")
    if (max_val-mean_val) > (min_val-mean_val):
        print("Big Profit")
    else:
        print("Big Loss")
elif kurto<0:
    print("Less distribution")
else:
    print("Normal distribution")
```

```python
import matplotlib.pyplot as plt
plt.plot(profit,color="green")
plt.title("Profit or Loss Distributions")
plt.xlabel("Profit values")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(profit,color="blue",edgecolor="green",bins=50)
plt.title("Profit or Loss Distribution")
plt.xlabel("Profit Values")
plt.ylabel("Frequency")
plt.show()
```

```python
data=pd.read_csv("tourism_dataset.csv")
data
```

```python
rate=data["Rating"]
r=rate.head(10)
country=data["Country"]
c=country.head(10)
print(r)
print(c)
```

```python
kurto=k(rate,fisher=True)
print("Kurtosis value:",kurto)
if kurto<0:
    print("Platykurtic Light tail low variance")
else:
    print("Not Platykurtic")
```

```python
plt.figure(figsize=(10,6))
plt.bar(r,c,color="red",width=0.5)
plt.title("Rating distribution")
plt.xlabel("Rating values")
plt.ylabel("Countries")
plt.tight_layout()
plt.show()
```
