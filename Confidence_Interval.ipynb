```python
import pandas as pd
import numpy as np
import scipy.stats as st
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("mobile_phone_screen_time_dataset.csv")
data
```

```python
screen_time=data["Daily_Screen_Time_Hours"].head(500)
screen_time
```

```python
result=st.t.interval(confidence=0.95,df=len(screen_time)-1,loc=np.mean(screen_time),scale=st.sem(screen_time))
print("95% Confidence Inteval:",result)
```

```python
loc=np.mean(screen_time)
print("Average :",loc)
```

```python
plt.figure(figsize=(6,4))
plt.hist(screen_time,bins=15,color="lightgreen",edgecolor="black")
plt.title("Daily Screen time of users")
plt.xlabel("Screen Time[Hours]")
plt.ylabel("No of Users")
plt.show()
```
