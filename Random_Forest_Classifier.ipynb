```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
```

```python
data=pd.read_csv("student_phone_addiction_affects_on_productivity(in).csv")
data
```

```python
print(data.columns)
```

```python
print(data.isnull().sum())
```

```python
data=data.dropna()
```

```python
X=data[["age", "daily_screen_time", "social_media_hours", "distraction_score", "productivity_score"]]
```

```python
print("\nMissing Values:")
print(X.isnull().sum())
```

```python
y=data["addiction_level"]
```

```python
print("\nMissing Values:")
print(y.isnull().sum())
```

```python
model=RandomForestClassifier(n_estimators=3, random_state=42)
model.fit(X, y)
```

```python
age=int(input("Enter your Age:"))
sct=int(input("Enter your Daily screen time:"))
smh=int(input("Enter your social media hours:"))
ds=int(input("Enter your Distraction score:"))
ps=int(input("Enter your Productivity score:"))
test_user=pd.DataFrame({"age":[age], "daily_screen_time":[sct], "social_media_hours":[smh], "distraction_score":[ds], "productivity_score":[ps]})

prediction=model.predict(test_user)
print("Predicted Result:", prediction[0])
```
