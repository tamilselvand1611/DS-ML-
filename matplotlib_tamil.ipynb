```python
import pandas as pd
data = pd.read_csv("Student_Marks.csv")
data
```

```python
m1=data.loc[0:5,"Marks"]
m2=data.loc[0:5,"time_study"]
```

```python
import matplotlib.pyplot as plt
plt.plot(m1,m2,color="blue",marker="s")
plt.title("Results")
plt.xlabel("marks")
plt.ylabel("study time")
plt.show()
```

```python
import numpy as np
plt.hist(m1,color="blue",bins=20,edgecolor="red")
plt.title("Results")
plt.xlabel("marks")
plt.ylabel("m")
plt.show()
```

```python
companies = ['Apple', 'Samsung', 'Xiaomi', 'Others']
market_share = np.array([30, 25, 20, 25])
plt.pie(market_share, labels=companies, autopct='%1.2f%%', startangle=90)
plt.title('Mobile Market Share')
plt.show()
```
