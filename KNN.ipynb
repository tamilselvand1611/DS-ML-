```python
data=pd.read_csv("Purchase.csv")
data
```

```python
X=data[["Salary"]]
y=data["Purchased"]
```

```python
x=np.array(X).reshape(-1, 1)
x
```

```python
y=np.array(y).reshape(-1, 1)
y
```

```python
model=KNeighborsClassifier(n_neighbors=3)
```

```python
model.fit(x, y)
```

```python
new_sal=[[63000]]
prediction=model.predict(new_sal)
```

```python
print("Predicted result:",prediction[0])
```
