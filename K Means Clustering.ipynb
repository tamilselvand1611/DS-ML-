```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data=pd.read_csv("Smartphone_Usage_And_Addiction_Analysis_7500_Rows.csv")
data
```

```python
print("\nMissing Values")
print(data.isnull().sum())
data.dropna(inplace=True)
data.drop_duplicates(inplace=True)
```

```python
X=data[["daily_screen_time_hours", "social_media_hours"]]
kmeans=KMeans(n_clusters=5, random_state=42)
clusters=kmeans.fit(X)
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(X["daily_screen_time_hours"], X["social_media_hours"], c=kmeans.labels_)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=300, color="Red", marker="X")
plt.xlabel("Daily Screen Time")
plt.ylabel("Social Media Hours")
plt.title("K Means Segmentation")
plt.grid(True)
plt.show()
```
