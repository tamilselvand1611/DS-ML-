```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt
```

```python
''' 95% defect-free items; probability exactly 9 good out of 10'''
p=0.95
n=10
k=9
```

```python
prob1=binom.pmf(k,n,p)
print("Probability of getting 9 defect-free items:",prob1)
```

```python
x=np.arange(0,n+1)
pmf_values=binom.pmf(x,n,p)
plt.bar(x,pmf_values,color="lightblue",edgecolor="black",label="Binomial PMF")
plt.bar(k,prob1,color="red",edgecolor="black",label=f"k = {k}")
plt.xlabel("Number of Exactly good ones")
plt.ylabel("Probability")
plt.title(f"Binomial distribution")
plt.legend()
plt.grid(axis="y",alpha=0.6)
plt.show()
```
