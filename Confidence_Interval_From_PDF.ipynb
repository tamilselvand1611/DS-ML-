```python
import scipy.stats as st
import math 
```

```python
'''50 students, mean=165, SD=10; 95% CI'''
n=50
z=1.96
mean=165
std=10
me=z*std/math.sqrt(n)
print("Margin of Error:",me)
ci=mean-me
print("95% CI starts from",ci,end=" ")
ci=mean+me
print("to",ci)
```

```python
result=st.norm.interval(confidence=0.95,loc=mean,scale=std/math.sqrt(n))
print("95% Confidence Inteval:",result)
```
