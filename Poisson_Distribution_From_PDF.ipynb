```python
from scipy.stats import poisson as p
import numpy as np
import matplotlib.pyplot as plt
```

```python
'''Website gets 6 visits/min; probability exactly 10 next minute'''
lam=6
k=10
prob1=p.pmf(k,lam)
print("Probability of getting visits next 10 mins:",prob1)
```

```python
x=np.arange(0,11)
fit=p.pmf(x,lam)
plt.bar(x,fit,color="lightblue",edgecolor="black")
plt.title("Poisson Distribution")
plt.xlabel("No of Website Visits")
plt.ylabel("Probability")
plt.show()
```
