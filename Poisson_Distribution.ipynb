```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson as p
```

```python
'''ATM Usage
An ATM is used by an average of 15 customers per hour.
Find the probability that exactly 18 customers use the ATM in one hour.'''
lam=15
k=18
pos=p.pmf(k,lam)
print("Probability of 18 customers use the ATM in one hour:",pos)
```

```python
x=np.arange(0,31)
fit=p.pmf(x,lam)
plt.bar(x,fit,color="lightblue",edgecolor="black")
plt.title("Poisson Distribution")
plt.xlabel("No of Customers use the ATM")
plt.ylabel("Probability")
plt.show()
```

```python
'''Typing Mistakes
A typist makes an average of 2 mistakes per page.
Find the probability of making exactly 5 mistakes on a page.'''
lam=2
k=5
pos=p.pmf(k,lam)
print("Probability of making exactly 5 mistakes:",pos)
```

```python
x=np.arange(0,11)
fit=p.pmf(x,lam)
plt.bar(x,fit,color="gold",edgecolor="black")
plt.title(f"Poisson Distribution")
plt.xlabel("No of mistakes on Page")
plt.ylabel("Probability")
plt.show()
```

```python
'''Rainfall Example
A city experiences an average of 4 heavy rain events per month.
Find the probability that there are exactly 6 heavy rain events this month.'''
lam=4
k=6
pos=p.pmf(k,lam)
print("Probability of 6 heavy rain events this month:",pos)
```

```python
x=np.arange(0,21)
fit=p.pmf(x,lam)
plt.bar(x,fit,color="green",edgecolor="black")
plt.title(f"Poisson Distribution")
plt.xlabel("No of Heavy rains this month")
plt.ylabel("Probability")
plt.show()
```
