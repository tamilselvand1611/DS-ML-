```python
import pandas as pd
import numpy as np
from scipy.stats import zscore
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("odb.csv")
data
```

```python
runs=data["Runs"]
```

```python
result=zscore(runs)
print("Z-Score:",result)
```

```python
data["Z_score"]=result
top=data.head(10)
top
```

```python
plt.figure(figsize=(10,5))
plt.bar(top["Player"],top["Z_score"],color="lightblue",edgecolor="black")
plt.title("Z-Score of Players")
plt.xticks(rotation=90)
plt.xlabel("Players")
plt.ylabel("Z-score")
plt.grid(axis="y",alpha=0.5)
plt.show()
```
